```python
import numpy as np
import csv
import random
import time

class Sudoku:
  """
  Sudoku Board Class
  """
  board: np.array([])
  solution: np.array([])

  def __init__(self, initBoard, solved):
    """
    Initialize Sudoku Board Object
    """
    initBoard = [int(x) for x in list(initBoard)]
    solved = [int(x) for x in list(solved)]
    self.board = np.reshape(initBoard, (9,9))
    self.solution = np.reshape(solved, (9,9))

  def printBoard(self):
    """
    Print Sudoku Board
    """
    for i in range(9):
      if i % 3 == 0 and i != 0:
        print("---------------------")
      for j in range(9):
        if j % 3 == 0 and j != 0:
          print("| ", end="")
        print(self.board[i][j], end=" ")
      print()
  def printSolution(self):
    """
    Print Sudoku Solution
    """
    for i in range(9):
      if i % 3 == 0 and i != 0:
        print("---------------------")
      for j in range(9):
        if j % 3 == 0 and j != 0:
          print("| ", end="")
        print(self.solution[i][j], end=" ")
      print()

  def updateCell(self, row, col, val):
    """
    Update Sudoku Cell
    """
    self.board[row][col] = val

  def checkArc():
    start = time.time()
    end = time.time()
    print("Time taken:", end-start)

  def checkForward():
    start = time.time()
    end = time.time()
    print("Time taken:", end-start)

#Main
data = []
solution = []
with open('sudoku.csv', newline='') as csvFile:
  reader = csv.reader(csvFile)
  next(reader)
  count = 10
  while count > 0:
    chance = random.random()
    if chance > 0.8:
      board = next(reader)
      data.append(board[0])
      solution.append(board[1])
      count -= 1
    else:
      next(reader)
boards = []
for i in range(len(data)):
  boards.append(Sudoku(data[i], solution[i]))

for i in range(len(data)):
  print("Board", i+1)
  boards[i].printBoard()
  print()
  print("Solution")
  boards[i].printSolution()
  print()
```
