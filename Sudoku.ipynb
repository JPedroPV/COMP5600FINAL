```python
import numpy as np
import csv
import random
import time

class Sudoku:
  """
  Sudoku Board Class
  """
  board: np.array([])
  solution: np.array([])

  def __init__(self, initBoard, solved):
    """
    Initialize Sudoku Board Object
    """
    initBoard = [int(x) for x in list(initBoard)]
    solved = [int(x) for x in list(solved)]
    self.board = np.reshape(initBoard, (9,9))
    self.solution = np.reshape(solved, (9,9))

  def printBoard(self):
    """
    Print Sudoku Board
    """
    for i in range(9):
      if i % 3 == 0 and i != 0:
        print("---------------------")
      for j in range(9):
        if j % 3 == 0 and j != 0:
          print("| ", end="")
        print(self.board[i][j], end=" ")
      print()
    print()

  def printSolution(self):
    """
    Print Sudoku Solution
    """
    for i in range(9):
      if i % 3 == 0 and i != 0:
        print("---------------------")
      for j in range(9):
        if j % 3 == 0 and j != 0:
          print("| ", end="")
        print(self.solution[i][j], end=" ")
      print()

  def updateCell(self, row, col, val):
    """
    Update Sudoku Cell
    """
    self.board[row][col] = val

  def checkCell(self, row, col):
    """
    Check if Sudoku Cell is Valid
    """
    val = self.board[row][col]
    if self.board[row][col] == 0:
      print("Empty Cell")
      return False
    for i in range(9):
      if self.board[row][i] == val and i != col:
        print("Row Violation")
        return False
      if self.board[i][col] == val and i != row:
        print("Column Violation")
        return False
    for i in range(3):
      for j in range(3):
        if self.board[row//3*3+i][col//3*3+j] == val and i != row and j != col:
          print("Subcell Violation")
          return False
    return True

  def checkArc():
    start = time.time()
    end = time.time()
    print("Time taken:", end-start)

  def checkForward():
    start = time.time()
    end = time.time()
    print("Time taken:", end-start)

  def checkMin():
    start = time.time()
    end = time.time()
    print("Time taken:", end-start)

  def checkMCV():
    start = time.time()
    end = time.time()
    print("Time taken:", end-start)

#Main
data = []
solution = []
with open('sudoku.csv', newline='') as csvFile:
  reader = csv.reader(csvFile)
  next(reader)
  count = 10
  while count > 0:
    chance = random.random()
    if chance > 0.0:
      board = next(reader)
      data.append(board[0])
      solution.append(board[1])
      count -= 1
    else:
      next(reader)
boards = []
for i in range(len(data)):
  boards.append(Sudoku(data[i], solution[i]))

# for i in range(len(data)):
#   print("Board", i+1)
#   boards[i].printBoard()
#   print()
#   print("Solution")
#   boards[i].printSolution()
#   print()

#checkCell() testing
help = boards[1]
help.printBoard()
print(help.checkCell(0, 0))
print()
help.updateCell(0, 0, 0)
help.printBoard()
print(help.checkCell(0, 0))
print()
help.updateCell(0, 0, 1)
help.printBoard()
print(help.checkCell(0, 0))
print()
help.updateCell(2, 5, 6)
help.printBoard()
print(help.checkCell(2, 5))
print()
help.updateCell(6, 1, 7)
help.printBoard()
print(help.checkCell(6, 1))
print()

```
